```python
#Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

print("Libraries imported successfully")
```

```python
df = pd.read_csv("../data/final_analytical_dataset.csv")

df.head()
```

```python
import os

os.listdir("../data")
```

```python
df.info()
```

```python
#Convert Date Columns
date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date',
    'shipping_limit_date'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors='coerce')

print("Data columns converted successfully")
```

```python
#Verify Date Types
df[date_columns].dtypes
```

```python
#Create order Date Features
df['order_year'] = df['order_purchase_timestamp'].dt.year
df['order_month'] = df['order_purchase_timestamp'].dt.month
df['order_month_name'] = df['order_purchase_timestamp'].dt.month_name()
df['order_day_name'] = df['order_purchase_timestamp'].dt.day_name()

df[['order_purchase_timestamp',
    'order_year',
    'order_month',
    'order_month_name',
    'order_day_name']].head()
```

```python
#Delivery Time Feature
df['delivery_time_days'] = (
    df['order_delivered_customer_date'] -
    df['order_purchase_timestamp']
).dt.days

df[['order_purchase_timestamp',
    'order_delivered_customer_date',
    'delivery_time_days']].head()
```

```python
#Create Revenue Features
df['total_sales'] = df['price']+df['freight_value']

df[['price',
    'freight_value',
    'total_sales']].head()
```

```python
#check revenue statistics
df['total_sales'].describe()
```

```python
#create delivery performance feature
df['delivery_delay_days'] = (
    df['order_delivered_customer_date'] -
    df['order_estimated_delivery_date']
).dt.days

df[['order_delivered_customer_date',
    'order_estimated_delivery_date',
    'delivery_delay_days']].head()
```

```python
#customer order count
customer_orders = df.groupby('customer_unique_id')['order_id'].count()

customer_orders.head()
```

```python
#Add Order Frequency Feature
df['customer_order_count'] = (
    df.groupby('customer_unique_id')['order_id']
    .transform('count')
)

df[['customer_unique_id',
    'customer_order_count']].head()
```

```python
#Customer Total Spending
df['customer_total_spending']=(
    df.groupby('customer_unique_id')['total_sales']
    .transform('sum')
)

df[['customer_unique_id',
    'customer_total_spending']].head()
```

```python
df.columns.tolist()
```

```python
df['total_sales'] = df['price'] + df['freight_value']
```

```python
df[['price','freight_value','total_sales']].head()
```

```python
#Delivery status
df['delivery_status']=np.where(
    df['delivery_delay_days'] <= 0,
    'On Time / Early',
    'Delayed'
)

df[['delivery_delay_days', 'delivery_status']].head()
```

```python
#Check Delivery Status Distribution
df['delivery_status'].value_counts()
```
